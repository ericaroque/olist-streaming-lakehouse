```python
%sql
SELECT 'orders' AS tabela, COUNT(*) AS total, _source_batch
FROM workspace.olist_landing.bronze_orders
GROUP by _source_batch

UNION ALL

SELECT 'items' AS tabela, COUNT(*) AS total, _source_batch
FROM workspace.olist_landing.bronze_items
GROUP by _source_batch

UNION ALL

SELECT 'payments' AS tabela, COUNT(*) AS total, _source_batch
FROM workspace.olist_landing.bronze_payments
GROUP by _source_batch;
```

```python
%sql
SELECT
    _source_batch,
    `_souerce_file`,
    COUNT(*) AS total
FROM workspace.olist_landing.bronze_items
GROUP BY
    _source_batch,
    `_souerce_file`
ORDER BY
    _source_batch,
    `_souerce_file`;
```

```python
%sql
SELECT
    order_id,
    order_item_id,
    COUNT(*) AS quantidade
FROM workspace.olist_landing.bronze_items
GROUP BY order_id, order_item_id
HAVING COUNT(*) > 1
ORDER BY quantidade DESC
LIMIT 20;
```
